```python
from yahpo_gym import BenchmarkSet

def main():
    b = BenchmarkSet("lcbench", instance="3945")
    

    cs = b.get_opt_space(drop_fidelity_params=True)
    cfg = cs.sample_configuration(1).get_dictionary()

    # LCBench epoch range: use a sensible fixed value first
    cfg["epoch"] = 51

    print("Sampled config:")
    print(cfg)

    out = b.objective_function(cfg)
    print("Output:")
    print(out)

if __name__ == "__main__":
    main()
```

```python

feature_keys = [
    'batch_size',
    'learning_rate',
    'max_dropout',
    'max_units',
    'momentum',
    'num_layers',
    'weight_decay',
]


b = BenchmarkSet("lcbench", instance="3945")

print("Targets:")
print(b.targets)
print("\n")

print("Instances:")
print(b.instances)
print("\n")


cs = b.get_opt_space(drop_fidelity_params=True)

for i, key in enumerate(feature_keys):
    print(f"Hyperparameter {i}: {key}")
    hp = cs.get_hyperparameter(key)
    print(hp)
    print(f"Lower bound: {hp.lower}")
    print(f"Upper bound: {hp.upper}")
    print(f"is log? {hp.log}")
    print("\n")
```

```python
from yahpo_gym import BenchmarkSet

def main():
    b = BenchmarkSet("lcbench", instance="3945")
    cs = b.get_opt_space(drop_fidelity_params=True)

    configs = cs.sample_configuration(8)
    if not isinstance(configs, list):
        configs = [configs]

    dicts = []
    for cfg in configs:
        d = cfg.get_dictionary()
        d["epoch"] = 51
        d["OpenML_task_id"] = "3945"
        dicts.append(d)

    out = b.objective_function(dicts)

    print(type(out))
    print(len(out) if isinstance(out, list) else "not list")
    print(out[0] if isinstance(out, list) else out)

if __name__ == "__main__":
    main()
```
